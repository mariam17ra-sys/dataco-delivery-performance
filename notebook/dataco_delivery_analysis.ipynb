```python
import pandas as pd

df = pd.read_csv("dataco_clean.csv")
df["order_date"] = pd.to_datetime(df["order_date"])
df["shipping_date"] = pd.to_datetime(df["shipping_date"])
```

```python
df_valid = df[df["delivery_status"] != "Shipping canceled"]

late_rate = (df_valid["delivery_status"] == "Late delivery").mean() * 100
print(round(late_rate, 1), "%")
```

```python
shipping_mode_rate = df_valid.groupby("shipping_mode")["delivery_status"].apply(
    lambda x: (x == "Late delivery").mean() * 100
).round(1).sort_values(ascending=False)

print(shipping_mode_rate)
```

```python
print(df_valid.groupby("shipping_mode")["days_for_shipment_scheduled"].mean().round(1))
```

```python
region_rate = df_valid.groupby("order_region")["delivery_status"].apply(
    lambda x: (x == "Late delivery").mean() * 100
).round(1).sort_values(ascending=False)

print(region_rate)
```

```python
df_valid["order_month"] = df_valid["order_date"].dt.to_period("M")

monthly_rate = df_valid.groupby("order_month")["delivery_status"].apply(
    lambda x: (x == "Late delivery").mean() * 100
).round(1)

print(monthly_rate)
```

```python
category_rate = df_valid.groupby("category_name")["delivery_status"].apply(
    lambda x: (x == "Late delivery").mean() * 100
).round(1).sort_values(ascending=False)

print(category_rate.head(10))
print("---")
print(category_rate.tail(10))
```

```python
df_valid["sales_bracket"] = pd.qcut(df_valid["sales"], q=4,
                                      labels=["Low", "Medium", "High", "Very High"])

sales_rate = df_valid.groupby("sales_bracket", observed=True)["delivery_status"].apply(
    lambda x: (x == "Late delivery").mean() * 100
).round(1)

print(sales_rate)
```

```python
segment_rate = df_valid.groupby("customer_segment")["delivery_status"].apply(
    lambda x: (x == "Late delivery").mean() * 100
).round(1).sort_values(ascending=False)

print(segment_rate)
```

```python
profit_by_status = df_valid.groupby("delivery_status")["order_profit_per_order"].mean().round(2)
print(profit_by_status)
```

```python
country_rate = df_valid.groupby("order_country")["delivery_status"].apply(
    lambda x: (x == "Late delivery").mean() * 100
).round(1).sort_values(ascending=False)

print(country_rate.head(10))
print("---")
print(country_rate.tail(10))
```

```python
country_counts = df_valid["order_country"].value_counts()
valid_countries = country_counts[country_counts >= 500].index

country_rate_filtered = df_valid[df_valid["order_country"].isin(valid_countries)].groupby("order_country")["delivery_status"].apply(
    lambda x: (x == "Late delivery").mean() * 100
).round(1).sort_values(ascending=False)

print(country_rate_filtered.head(10))
print("---")
print(country_rate_filtered.tail(10))
```
